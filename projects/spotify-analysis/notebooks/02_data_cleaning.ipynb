# Data Cleaning

## Objective
The objective of this notebook is to prepare the dataset for the exploratory analysis.

The cleaning process includes:
- Removing observations with missing values in essential metadata.
- Removing duplicated songs based on `track_id`.
- Verifying the consistency of the cleaned dataset.
- Saving the cleaned dataset for subsequent analysis.

```python
import pandas as pd
path = "../data/raw/spotify_songs.csv"

df_original = pd.read_csv(path)
```

### Removing missing values

```python
rows_before=len(df_original)

df = df_original.dropna()
rows_after=len(df)
print(f"Rows before: {rows_before}\nRows after: {rows_after}\nRows dropped: {rows_before-rows_after}")
```

### Remove duplicated songs
The dataset contains multiple occurrences of the same song because a track can belong to multiple playlists.
Since the objective of this project is to analyse songs rather than playlists, duplicated `track_id` values are removed, while keeping the first occurrence.


```python
rows_before=len(df)
df = df.drop_duplicates(subset ='track_id', keep = 'first')
rows_after=len(df)

print(f"Rows before: {rows_before}\nRows after: {rows_after}\nDuplicates removed: {rows_before-rows_after}")
```

```python
missing_track_ids = df_original[df_original.isna().any(axis=1) & df_original.duplicated(subset='track_id', keep='first')]["track_id"]
print(missing_track_ids)
```

### Save cleaned dataset

```python
output_path = "../data/processed/spotify_songs_cleaned.csv"
df.to_csv(output_path, index=False)
```

The following cleaning steps were performed:
- Removed **5 observations** containing missing values.

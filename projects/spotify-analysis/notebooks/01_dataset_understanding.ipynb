# Spotify Songs Analysis

## Objective
The objective of this project is to analyze which characteristics are associated with the popularity of songs on Spotify.


## Dataset
Source: Kaggle - Spotify Songs Dataset

Each row represents a song, containing information about:
- track metadata
- artist information
- audio features
- popularity score


## Main questions
1. Which variables are associated with song popularity?
2. Are there differences between musical genres?
3. How has song popularity changed over time?
4. What characteristics do the most popular songs share?
5. Are there interesting correlations between audio features?
6. What recommendations can I provide to the product team based on the analysis?

## Summary
The initial inspection of the dataset led to the following findings:

- The dataset contains **32833 observations** and **23 variables**.
- Each row represents a **song occurrence within a Spotify playlist**, nor necessarily a unique song.
- Five observations contain missing values in track metadata (`track_name`, `track_artist` and `track_albun_name`).
- No fully duplicated rows were found.
- A total of **4477 duplicated `track_id` values** were identified, indicating that the same song can appear in multiple playlists.
- The next step is to clean the dataset by removing missing values and deciding how to handle duplicated songs.
